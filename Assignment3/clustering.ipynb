#Moons Dataset (DBSCAN excels): Non-linear, noisy clusters.

Blobs with Varying Density (DBSCAN struggles): Clusters with different standard deviations.


```python
from sklearn.datasets import make_moons, make_blobs
import matplotlib.pyplot as plt

# Dataset 1: Moons (DBSCAN excels)
X_moons, y_moons = make_moons(n_samples=300, noise=0.05, random_state=42)

# Dataset 2: Blobs with varying density (DBSCAN struggles)
X_blobs, y_blobs = make_blobs(n_samples=300, centers=3, cluster_std=[1.0, 2.5, 0.5], random_state=42)


#APPLYING CLUSTERING ALGORITHMS
from sklearn.cluster import DBSCAN, KMeans , AgglomerativeClustering 

# DBSCAN on Moons
dbscan_moons = DBSCAN(eps=0.2, min_samples=5)
dbscan_moons.fit(X_moons)
# KMeans on Moons
kmeans_moons = KMeans(n_clusters=2, random_state=42)
kmeans_moons.fit(X_moons)
#hierarchical clustering on Moons
from sklearn.cluster import AgglomerativeClustering
agglo_moons = AgglomerativeClustering(n_clusters=2)
agglo_moons.fit(X_moons)

# DBSCAN on Blobs
dbscan_blobs = DBSCAN(eps=0.5, min_samples=5)
dbscan_blobs.fit(X_blobs)
# KMeans on Blobs
kmeans_blobs = KMeans(n_clusters=3, random_state=42)    
kmeans_blobs.fit(X_blobs)
# Hierarchical clustering on Blobs
agglo_blobs = AgglomerativeClustering(n_clusters=3)
agglo_blobs.fit(X_blobs)


#PLOTTING RESULTS FOR MOONS DATASET
plt.figure(figsize=(15, 5))

# Plot DBSCAN
plt.subplot(1, 3, 1)
plt.scatter(X_moons[:, 0], X_moons[:, 1], c=dbscan_moons.labels_, cmap='viridis')
plt.title("DBSCAN (Moons)")

# Plot k-Means
plt.subplot(1, 3, 2)
plt.scatter(X_moons[:, 0], X_moons[:, 1], c=kmeans_moons.labels_, cmap='viridis')
plt.title("k-Means (Moons)")

# Plot Hierarchical
plt.subplot(1, 3, 3)
plt.scatter(X_moons[:, 0], X_moons[:, 1], c=agglo_moons.labels_, cmap='viridis')
plt.title("Hierarchical (Moons)")

plt.tight_layout()
plt.show()



# PLOTTING RESULTS FOR BLOBS DATASET
plt.figure(figsize=(15, 5))

# Plot DBSCAN
plt.subplot(1, 3, 1)
plt.scatter(X_blobs[:, 0], X_blobs[:, 1], c=dbscan_blobs.labels_, cmap='viridis')
plt.title("DBSCAN (Blobs)")

# Plot k-Means
plt.subplot(1, 3, 2)
plt.scatter(X_blobs[:, 0], X_blobs[:, 1], c=kmeans_blobs.labels_, cmap='viridis')
plt.title("k-Means (Blobs)")

# Plot Hierarchical
plt.subplot(1, 3, 3)
plt.scatter(X_blobs[:, 0], X_blobs[:, 1], c=agglo_blobs.labels_, cmap='viridis')
plt.title("Hierarchical (Blobs)")

plt.tight_layout()
plt.show()
```
